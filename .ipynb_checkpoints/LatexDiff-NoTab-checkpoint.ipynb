```python
import re
```

```python
rep="/Users/gilles/sDrive/Recherche/Boye/HDR/Memoire/"
```

```python
def supTab(nTex):
    with open(rep+nTex,"r") as inFile:
        tex=inFile.readlines()
    
    texte_latex="".join(tex)

    """
    Extrait les tables du texte et les remplace par des marqueurs uniques.
    """
    tables_stockees = {}
    
    # Regex pour capturer les environnements table / table* / tabular / tabular*
    # re.DOTALL (ou flag s) permet à '.' de capturer aussi les sauts de ligne
    pattern_table = r'\\begin{table}(.|\n)+?\\end{table}'

    def remplacer(match):
        index = len(tables_stockees)
        cle = f"%___TABLE_PLACEHOLDER_{index}___%"
        tables_stockees[cle] = match.group(0)
        return cle

    # Remplace chaque table trouvée par sa clé
    texte_masque = re.sub(pattern_table, remplacer, texte_latex, flags=re.DOTALL)
    
    return texte_masque, tables_stockees
```

```python
nOld="2-PerspectiveRealiste-old.tex"
nNew="2-PerspectiveRealiste-new.tex"
oldTex,oldTabs=supTab(nOld)
with open(rep+nOld.replace(".tex","-noTab.tex"),"w") as outFile:
        outFile.write(oldTex)
newTex,newTabs=supTab(nNew)
with open(rep+nNew.replace(".tex","-noTab.tex"),"w") as outFile:
        outFile.write(newTex)
```

```python
def restaurer_tables(texte_traite, tables_stockees):
    """
    Reinjecte les tables d'origine à la place des marqueurs.
    """
    texte_restaure = texte_traite
    for cle, table_originale in tables_stockees.items():
        texte_restaure = texte_restaure.replace(cle, table_originale)
    return texte_restaure
```

```python
print(rep+nNew.replace("new","noTab"))
with open(rep+nNew.replace("new","noTab"),"r") as inFile:
        noTabTex=inFile.read()
tabTex=restaurer_tables(noTabTex,newTabs)
with open(rep+nNew.replace("new","Tab"),"w") as outFile:
        outFile.write(tabTex)
```
