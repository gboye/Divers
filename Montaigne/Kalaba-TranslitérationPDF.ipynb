- importer pypdf pour manipuler directement le fichier
- faire deux tables
    1. une table pour les polygraphes => replace
    1. une table pour les monographes => maketrans
- lire le fichier
    - séparer les lignes
    - traduire chaque ligne

```python
import pypdf
```

```python
repKalabia="/Users/gilles/sDrive/Cours/Bordeaux/L1-LinguistiqueGenerale/00-ProjetKalaba/25-K5/"
nKalabia="25-K5-Devoir.pdf"
# repKalabia="/Users/gilles/sDrive/Cours/Bordeaux/L1-LinguistiqueGenerale/00-ProjetKalaba/KalabasPrecedents/23-K5/"
# nKalabia="23-K5-Devoir.pdf"
```

```python
reader = pypdf.PdfReader(repKalabia+nKalabia)
```

```python
trans2={"ng":"N",
        "sh":"S",
        "th":"T",
        "dh":"D",
        "ca":"a",
        "ce":"e",
        "ci":"i",
        "co":"o",
        "cu":"u",
        "R":"r",        
        "j":"Z",        
       }
transK="rljwptkbdgmnNfsTSvzDZaeiou"
transA="ptkbdgmnNfsTSvzDZrljwiueoa"
trans1=str.maketrans(transK,transA)
transSampa="SZNTDRy"
transAPI="ʃʒŋθðrj"
trans=str.maketrans(transSampa,transAPI)
```

```python
def traduire(ligne):
    result=ligne
    for k,v in trans2.items():
        result=result.replace(k,v)
    result=result.translate(trans1).translate(trans)
    return result
```

```python
nPages=len(reader.pages)
for p in range(nPages):
    print(p+1)
    lignes=reader.pages[p].extract_text().split("\n")
    for ligne in lignes[:1]: # Seulement la première ligne de chaque page
        print(ligne)
        print(traduire(ligne))
        print()
```
