# Générer des questions GapSelect pour les gloses
- de noms orthographiques

```python
from lxml import etree
```

```python
def gen_ortho(mots,glose,phrase):
    '''
    ajoute une question au quiz
    
    mots contient les différents lemmes à proposer sous forme de liste
    glose contient les éléments de la glose sous forme de liste
    phrase contient la phrase de contexte

    forme cherche le mot tel qu'il apparaît dans la phrase
    '''
    
    forme=[mot for mot in mots if mot in phrase.split(" ")]
    if len(forme)==1:
        forme=forme[0]
        print(forme)
    else:
        forme=None
        print("Problème avec la forme du mot dans la phrase")
        return
        
    '''
    calcul le nombre de champs de la glose pour le CDATA
    '''    
    champs=[f"[[{i+1}]]" for i in range(len(glose))]
    gaps="".join(champs)+"</span></p>"
    
    # 2. Création de la question gapselect
    question = etree.SubElement(quiz, "question", type="gapselect")
    
    # Nom de la question
    name = etree.SubElement(question, "name")
    etree.SubElement(name, "text").text = "".join(glose)
    
    # 3. Gestion du texte de la question avec CDATA natif
    questiontext = etree.SubElement(question, "questiontext", format="html")
    text_element = etree.SubElement(questiontext, "text")
    
    # Contenu HTML exact adapté pour "chiens"
    # html_lines
    html_content = "".join([f'<p dir="ltr" style="text-align: left;"><b>{forme}</b> dans {phrase}</p>',
    f'<p dir="ltr" style="text-align: left;"><span style="font-size: calc({{0.90375rem}}{{0.045vw}});">',
    # f'[[1]][[2]][[3]][[4]][[5]][[6]][[7]]</span></p>'
    gaps
                           ])

    # Encapsulation automatique et propre en CDATA
    text_element.text = etree.CDATA(html_content)
    
    # 4. Paramètres et feedbacks standards
    etree.SubElement(question, "generalfeedback", format="html").text = ""
    etree.SubElement(question, "defaultgrade").text = "1"
    etree.SubElement(question, "penalty").text = "0.3333333"
    etree.SubElement(question, "hidden").text = "0"
    etree.SubElement(question, "idnumber").text = ""
    etree.SubElement(question, "shuffleanswers").text = "1"
    
    correct = etree.SubElement(question, "correctfeedback", format="html")
    etree.SubElement(correct, "text").text = "Votre réponse est correcte."
    
    partially = etree.SubElement(question, "partiallycorrectfeedback", format="html")
    etree.SubElement(partially, "text").text = "Votre réponse est partiellement correcte."
    
    incorrect = etree.SubElement(question, "incorrectfeedback", format="html")
    etree.SubElement(incorrect, "text").text = "Votre réponse est incorrecte."
    
    etree.SubElement(question, "shownumcorrect")
    
    # 5. Définition des options de sélection (Selectoptions)
    # Configuration grammaticale pour "chiens" : Nom (N), Masculin (M), Pluriel (PL)
    options=[]
    for g in glose:
        if g in mots:
            options.append((g,"1"))
        elif g in separateurs:
            options.append((g,"2"))
        elif g in proprietes:
            options.append((g,"3"))
    for m in mots: 
        if m!=glose[0]:
            options.append((m,"1"))
    for separateur in separateurs:
        if separateur not in [option[0] for option in options]:
            options+=[(separateur,"2")]
    for propriete in proprietes:
        if propriete not in [option[0] for option in options]:
            options+=[(propriete,"3")]
    for text, group in options:
        opt = etree.SubElement(question, "selectoption")
        etree.SubElement(opt, "text").text = text
        etree.SubElement(opt, "group").text = group
        
    return       
```

```python
proprietes=["M","F","N","SG","DU","PL"]
separateurs="()[]-."


if __name__ == "__main__":
    # 1. Création de la racine <quiz>
    quiz = etree.Element("quiz")
    category_question = etree.SubElement(quiz, "question", type="category")
    category_node = etree.SubElement(category_question, "category")
    etree.SubElement(category_node, "text").text = "Gloses"

    mots=["vacances","vacance"]
    glose=["vacances","(","F",".","PL",")"]
    phrase="C'est les vacances !"
    phon="selevakãs"
    morph="vakans"

    gen_ortho(mots,glose,phrase)


    mots=["bassin","bassins","bassine","bassines"]
    glose=["bassine","(","F",")","[","PL","]"]
    phrase="Les bassines ont débordé."
    phon="lebasinõzdeborde"
    morph="basin"

    gen_ortho(mots,glose,phrase)
    
    mots=["cheval","chevaux"]
    glose=["cheval","(","M",")","-","PL"]
    phrase="Les chevaux courent."
    phon="leS@vokur"
    morph="S@v-o"

    gen_ortho(mots,glose,phrase)

    
    
    # Écriture du fichier avec Pretty Print natif et déclaration XML
    output_filename = "Gapselect-Gloses.xml"
    
    # lxml gère l'encodage et le formatage de manière très propre
    doc = etree.ElementTree(quiz)
    doc.write(
        output_filename, 
        pretty_print=True, 
        xml_declaration=True, 
        encoding="UTF-8"
    )
    
    print(f"Fichier XML généré avec succès via lxml : {output_filename}")
```
