```python
import glob,shutil,re
```

```python
base="/Volumes/BroadExt/Gilles/Downloads/1LNSU4-Dossier Kalabia-46492-2/"
reps=glob.glob(base+"*/")
for rep in reps:
    # print(rep)
    m=re.search("^/.*/Kalabia-(\d\d)",rep)
    groupe=m.group(1)
    files=glob.glob(rep+"*.*")
    if len(files)==1:
        shutil.copy(files[0],base+"Kalabia-%s.pdf"%groupe)
    else:
        print("plusieurs fichiers")
```
